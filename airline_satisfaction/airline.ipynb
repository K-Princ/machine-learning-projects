```python
df_full_train.columns = df_full_train.columns.str.lower().str.replace(' ', '_')
df_full_train.columns = df_full_train.columns.str.lower().str.replace('/', '_')

categorical_columns = list(df_full_train.dtypes[df_full_train.dtypes == 'str'].index)

for c in categorical_columns:
    df_full_train[c] = df_full_train[c].str.lower().str.replace(' ', '_')
```

```python
df_test.columns = df_test.columns.str.lower().str.replace(' ', '_')
df_test.columns = df_test.columns.str.lower().str.replace('/', '_')

categorical_columns = list(df_test.dtypes[df_test.dtypes == 'str'].index)

for c in categorical_columns:
    df_test[c] = df_test[c].str.lower().str.replace(' ', '_')
```

```python
df_test.arrival_delay_in_minutes = df_test.arrival_delay_in_minutes.fillna(df_test.arrival_delay_in_minutes.median())
```

```python
df_full_train.satisfaction = df_full_train.satisfaction.astype(int)
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_train, df_val = train_test_split(df_full_train, test_size = 0.2, random_state = 42)
```

```python
len(df_train), len(df_val)
```

```python
y_train = df_train.satisfaction.values
y_val = df_val.satisfaction.values
y_full_train = df_full_train.satisfaction.values
```

```python
df_full_train = df_full_train.reset_index(drop = True)
df_train = df_train.reset_index(drop = True)
df_test = df_test.reset_index(drop = True)
```

```python
del df_full_train['satisfaction']
del df_train['satisfaction']
del df_val['satisfaction']
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.feature_extraction import DictVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score
```

```python
train_dicts = df_train.to_dict(orient = "records")
```

```python
dv = DictVectorizer(sparse = False)
X_train = dv.fit_transform(train_dicts)
```

```python
lr = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        solver='lbfgs',
        C=1.0,
        max_iter=300,
        random_state=42
    )
)
lr.fit(X_train, y_train)
```

```python
val_dicts = df_val.to_dict(orient = "records")
X_val = dv.transform(val_dicts)
```

```python
y_pred = lr.predict_proba(X_val)[:,1]
```

```python
roc_auc_score(y_val, y_pred)
```

```python
dicts_full_train = df_full_train.to_dict(orient='records')

dv = DictVectorizer(sparse=False)
X_full_train = dv.fit_transform(dicts_full_train)
lr.fit(X_full_train, y_full_train)

dicts_test = df_test.to_dict(orient='records')
X_test = dv.transform(dicts_test)
```

```python
predictions = lr.predict(X_test)
```

```python
submission = pd.DataFrame({
    'id': df_test['id'],
    'satisfaction': predictions
})

submission.to_csv('submission.csv', index=False)
```
